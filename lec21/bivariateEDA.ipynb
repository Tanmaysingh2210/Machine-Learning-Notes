```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os

dir_add = os.getcwd()
train_path = os.path.join(dir_add, "train.csv")
titanic = pd.read_csv(train_path)

titanic.head()
```

```python
# numerical - numerical columns

# scatter plot 
sns.scatterplot(x = titanic['Age'], y = titanic['Fare'], hue=titanic['Sex']) #no relationship in Age and Fair but still we are trying 
```

```python
# Numerical - categorical

# BAR plot

sns.barplot(x = titanic['Pclass'],y= titanic['Age'], hue=titanic['Sex'])
```

```python
# Numerical - categorical

# BAR plot

sns.barplot(x = titanic['Pclass'],y= titanic['Fare'])
```

```python
# BOX PLOT(Numerical - Categorical)

sns.boxplot(x= titanic['Sex'], y = titanic['Age'])
```

```python
# distplot (Numerical - categorical)

# sns.distplot(x= titanic['Age'])

# making hue by myself
sns.distplot(titanic[titanic['Survived']==0]['Age'], hist=False) #died - blue
sns.distplot(titanic[titanic['Survived']==1]['Age'], hist = False) #survived - orange
```

```python
# Categorical - Categorical Analysis

# HeatMap 
# pd.crosstab(titanic['Pclass'], titanic['Survived'])
sns.heatmap(pd.crosstab(titanic['Pclass'], titanic['Survived']))
```

```python
titanic.groupby('Pclass')['Survived'].mean()*100
```

```python
(titanic.groupby('Sex')['Survived'].mean()*100).plot(kind='bar')
```

```python
# ClusterMap (categorical - categorical)
pd.crosstab(titanic['SibSp'], titanic['Survived'])
```

```python
sns.clustermap(pd.crosstab(titanic['Parch'], titanic['Survived']))
```
