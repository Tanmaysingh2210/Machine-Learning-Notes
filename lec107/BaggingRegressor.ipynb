```python
import numpy as np
import pandas as pd

data_url = "http://lib.stat.cmu.edu/datasets/boston" # load boston dataset
raw_df = pd.read_csv(data_url, sep="\s+", skiprows=22, header=None)
data = np.hstack([raw_df.values[::2, :], raw_df.values[1::2, :2]])
target = raw_df.values[1::2, 2]
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.neighbors import KNeighborsRegressor 
from sklearn.model_selection import GridSearchCV 
from sklearn.tree import DecisionTreeRegressor
from sklearn.metrics import r2_score
```

```python
from sklearn.model_selection import train_test_split
xtrain, xtest, ytrain, ytest = train_test_split(data, target, test_size=0.2, random_state=2)
```

```python
lr = LinearRegression()
dt = DecisionTreeRegressor()
knn = KNeighborsRegressor()
```

```python
lr.fit(xtrain, ytrain)
dt.fit(xtrain, ytrain)
knn.fit(xtrain, ytrain)
```

```python
print("LR", r2_score(ytest, lr.predict(xtest)))
print("DT", r2_score(ytest, dt.predict(xtest)))
print("KNN", r2_score(ytest, knn.predict(xtest)))
```

```python
from sklearn.ensemble import BaggingRegressor
bag = BaggingRegressor()
bag.fit(xtrain, ytrain)
```

```python
print("Trianing coef" , bag.score(xtrain, ytrain))
print("Test coef" , bag.score(xtest, ytest))
```
