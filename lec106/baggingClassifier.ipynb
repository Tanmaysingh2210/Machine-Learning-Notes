```python
from sklearn.datasets import make_classification
from sklearn.metrics import accuracy_score
from sklearn.ensemble import BaggingClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.svm import SVC
from sklearn.model_selection import train_test_split
```

```python
x,y = make_classification(n_samples=10000, n_features=10, n_informative=3)
xtrain, xtest, ytrain, ytest = train_test_split(x,y,test_size=0.2, random_state=42)
```

```python
dt = DecisionTreeClassifier(random_state=42)
dt.fit(xtrain, ytrain)

accuracy_score(ytest, dt.predict(xtest))
```

Bagging


```python
bag = BaggingClassifier(
    estimator=DecisionTreeClassifier(),
    n_estimators=500,
    max_samples=0.25,
    bootstrap=True,
    random_state=42,
)
```

```python
bag.fit(xtrain, ytrain)
```

```python
accuracy_score(ytest , bag.predict(xtest))
```

```python
bag.estimators_samples_
```

```python
bag.estimators_features_
```

Bagging with SVM


```python
bag = BaggingClassifier(
    estimator=SVC(),
    n_estimators=500,
    max_samples=0.25,
    bootstrap=True,
    random_state=42
)
```

```python
bag.fit(xtrain, ytrain)
```

```python
accuracy_score(ytest, bag.predict(xtest))
```

Pasting

```python
bag = BaggingClassifier(
    estimator=DecisionTreeClassifier(),
    n_estimators=500,
    max_samples=0.25,
    bootstrap=False,   #sampling without reokacement 
    random_state=42,
    verbose=1,  
    n_jobs=-1  #devide the task in all cores of cpu in my system
)
```

```python
bag.fit(xtrain, ytrain)
accuracy_score(ytest, bag.predict(xtest))
```

Random Subsoaces

```python
bag = BaggingClassifier(
    estimator=DecisionTreeClassifier(),
    n_estimators=500,
    max_samples=1.0,
    bootstrap=False,
    max_features=0.5, # column sampling 
    bootstrap_features=True,
    random_state=42,
    n_jobs=-1
)
```

```python
bag.fit(xtrain, ytrain)
accuracy_score(ytest, bag.predict(xtest))
```

```python
bag.estimators_samples_
```

```python
bag.estimators_features_
```

Random patches

```python
bag = BaggingClassifier(
    estimator=DecisionTreeClassifier(),
    n_estimators=500,
    max_samples=0.25, #row sampling 
    bootstrap=True,
    max_features=0.5, # column sampling 
    bootstrap_features=True,
    random_state=42,
    n_jobs=-1
)
```

```python
bag.fit(xtrain, ytrain)
accuracy_score(ytest, bag.predict(xtest))
```

OOB Score - around 37% rows are not used to train any decision tree =. therefore they are called "Out Of Bag"  

```python
bag = BaggingClassifier(
    estimator=DecisionTreeClassifier(),
    n_estimators=500,
    max_samples=0.25,
    bootstrap=True,
    oob_score=True,
    random_state=42
)
```

```python
bag.fit(xtrain, ytrain)
```

```python
bag.oob_score_  # tell about accuracy 
```

```python
accuracy_score(ytest, bag.predict(xtest))
```

BAGGING TIPS:

    Bagging generally gives better results than Pasting.


    Good results come around the 25% to 50% row sampling mark.


    Random Patches and subspaces should be used while dealing with high dimensional data.


    To find the correct hyperparameter values we can do GridSearchCV/RandomSearchCV.
