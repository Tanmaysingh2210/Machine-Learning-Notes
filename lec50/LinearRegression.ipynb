```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import os

cwd = os.getcwd()
path = os.path.join(cwd, "placements.csv")
df = pd.read_csv(path)
df.head()
```

```python
plt.scatter(df['cgpa'], df['package'])
plt.xlabel('CGPA')
plt.ylabel('Package(in LPA)')
plt.show()
```

```python
# data is sort of linear 
x = df.iloc[:, 0:1]
y= df.iloc[:, 1:]
y.head()
```

```python
from sklearn.model_selection import train_test_split
x_train, X_test , Y_train, Y_test = train_test_split(x,y,test_size=0.2, random_state=2)
```

```python
from sklearn.linear_model import LinearRegression
lr = LinearRegression()
```

```python
lr.fit(x_train,Y_train)
```

```python
X_test
```

```python
Y_test
```

```python
X_test.iloc[0].values.reshape(1,1) # cgpa
```

```python
lr.predict(X_test.iloc[2].values.reshape(1,1))
```

```python
plt.scatter(df['cgpa'], df['package'])
plt.plot(X_test, lr.predict(X_test), color="red") #best fit line plot
plt.xlabel('CGPA')
plt.ylabel('Package(in LPA)')
plt.show()
```

```python
m = lr.coef_ #slope of line
m
```
